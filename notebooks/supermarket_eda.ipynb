# Data Cleaning

```python
df.fillna(0,inplace=True)
```

```python
df.drop_duplicates(inplace=True)
```

# Feature Engineering

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True)

df['Ship Date'] = pd.to_datetime(df['Ship Date'], dayfirst=True)
```

```python
df['Year'] = df['Order Date'].dt.year

df['Month'] = df['Order Date'].dt.month_name()

df['Day'] = df['Order Date'].dt.day_name()
```

```python
df.head()
```

# Visualization

```python
plt.figure(figsize=(10,5))

sns.histplot(df['Sales'], bins=30, kde=True)

plt.title("Sales Distribution")

plt.xlabel("Sales")

plt.ylabel("Count")

plt.savefig("../images/sales_distribution.png", bbox_inches='tight')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(x='Region', y='Sales', data=df)

plt.title("Region Wise Sales")

plt.xlabel("Region")

plt.ylabel("Sales")

plt.savefig("../images/region_sales.png", bbox_inches='tight')

plt.show()
```

```python
plt.figure(figsize=(8,5))

ax = sns.countplot(x='Category', data=df)

for p in ax.patches:
    count = int(p.get_height())
    
    ax.annotate(f'{count}',
                (p.get_x() + p.get_width() / 2., count),
                ha='center', va='bottom')

plt.title("Product Categories")

plt.xlabel("Category")

plt.ylabel("Number of Products")

plt.savefig("../images/category_analysis.png", bbox_inches='tight')

plt.show()
```

```python
plt.figure(figsize=(12,6))

ax = sns.countplot(y='Sub-Category', data=df)

for p in ax.patches:
    count = int(p.get_width())
    
    ax.annotate(f'{count}',
                (count, p.get_y() + p.get_height()/2),
                ha='left', va='center')

plt.title("Sub Categories")

plt.xlabel("Number of Products")

plt.ylabel("Sub-Category")

plt.savefig("../images/subcategory_analysis.png", bbox_inches='tight')

plt.show()
```

```python
monthly_sales = df.groupby('Month')['Sales'].sum()

plt.figure(figsize=(12,6))

monthly_sales.plot(kind='line', marker='o')

plt.title("Monthly Sales Trend")

plt.xlabel("Month")

plt.ylabel("Sales")

plt.savefig("../images/monthly_sales.png", bbox_inches='tight')

plt.show()
```

# Business Insights

1. Office Supplies category contains the highest number of products with more than 5900 entries.

2. Technology category has fewer products compared to Office Supplies but contributes strongly to business value.

3. The East and South regions show slightly higher average sales compared to Central and West regions.

4. Binders and Paper are the most frequently sold sub-categories.

5. Copiers and Machines have the lowest product counts among all sub-categories.

6. Monthly sales fluctuate significantly, with some months showing very high sales peaks.

7. Sales distribution is highly right-skewed, indicating that most transactions are low-value while a few transactions are extremely high-value.
